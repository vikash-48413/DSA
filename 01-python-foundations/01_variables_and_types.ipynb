## Q019 — Advanced

**Given n=987654, calculate digit count, first digit, last digit, digit sum, and reverse using arithmetic only.**

**Your solution:**

```python
# Write your code here
n = 987654
temp  = n
counter = 0
digit_sum = 0
reverse = 0
last_digit = n % 10

while temp > 0:
    r = temp % 10
    reverse = reverse * 10 + r
    temp = temp // 10
    counter += 1
    digit_sum += r

print("Counter:", counter)
print("Sum:", digit_sum)
print("Reverse:", reverse)

print("First Digit:", r)
print("Last Digit:", last_digit)
```
