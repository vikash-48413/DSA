## 1. Implementation From Scratch (Pure Python)
Implement the core concept below without using high-level external libraries.

```python
from typing import List, Optional

def calculate_embedding_mean(vectors: List[List[float]]) -> Optional[float]:
    if not vectors:
        return None
    total_sum = sum(sum(v) for v in vectors)
    total_count = sum(len(v) for v in vectors)
    return total_sum / total_count if total_count > 0 else 0.0

sample_vectors = [[1.0, 2.0], [3.0, 4.0]]
print('Mean:', calculate_embedding_mean(sample_vectors))
```

## 2. Unit Testing & Edge Case Verification
Run the test assertions below to verify correctness of your from-scratch implementation.

```python
assert calculate_embedding_mean([[1.0, 2.0]]) == 1.5
assert calculate_embedding_mean([]) is None
print('All variable and type tests passed!')
```
