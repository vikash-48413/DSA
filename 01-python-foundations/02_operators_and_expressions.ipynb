### Question 9: Midpoint Index for Binary Search
**Prompt:** Calculate the midpoint index between `low = 3` and `high = 11` using integer arithmetic to prevent fractional values.


```python
low = 3
high = 11
# In Python: (low + high) // 2 or low + (high - low) // 2 (to avoid overflow in languages with fixed int sizes)
mid = low + (high - low) // 2
print('Midpoint index:', mid)
```

### Question 11: Chained Comparisons
**Prompt:** Predict the output of the chained comparison: `1 < 2 < 3 == 3 < 4`.


```python
# Chained comparisons expand as: (1 < 2) and (2 < 3) and (3 == 3) and (3 < 4)
result = 1 < 2 < 3 == 3 < 4
print('Result:', result)
```

### Question 12: Comparison Precedence Pitfall
**Prompt:** Predict the output of: `10 > 20 == False`. (Explain operator precedence between `>` and `==`).


```python
# Both > and == have the same comparison precedence and chain together!
# It evaluates as: (10 > 20) and (20 == False)
# 10 > 20 is False, so short-circuit returns False (it does NOT evaluate 10 > (20 == False))
result = 10 > 20 == False
print('10 > 20 == False evaluates to:', result)
```

### Question 13: Value Equality vs Object Identity (`==` vs `is`)
**Prompt:** Given `a = [1, 2]` and `b = [1, 2]`, evaluate `a == b` vs `a is b`.


```python
a = [1, 2]
b = [1, 2]

print('a == b (value check):', a == b)
print('a is b (memory identity check):', a is b)
print('id(a):', id(a), '| id(b):', id(b))
```

### Question 14: Integer Caching Pitfall
**Prompt:** Given `x = 256` and `y = 256`, evaluate `x is y`. Repeat for `x = 1000` and `y = 1000` and explain Python's small integer caching.


```python
# Python caches small integers in the range [-5, 256]
x1, y1 = 256, 256
print('256 is 256:', x1 is y1)

x2 = int('1000')
y2 = int('1000')
print('1000 is 1000 (dynamic):', x2 is y2)
```

### Question 23: Short-Circuiting with `and`
**Prompt:** Predict the output without running the code: `False and (1 / 0)`. Why does it not raise a `ZeroDivisionError`?


```python
# Because the first operand is False, Python short-circuits and never evaluates the right side!
result = False and (1 / 0)
print('Result:', result)
```

### Question 24: Short-Circuiting with `or`
**Prompt:** Predict the output: `True or (1 / 0)`.


```python
# Because the first operand is True, Python short-circuits and returns True immediately!
result = True or (1 / 0)
print('Result:', result)
```

### Question 27: De Morgan's Law Simplification
**Prompt:** Simplify the Boolean expression: `not (A and not B)`.


```python
# De Morgan's Law: not (A and B) = (not A) or (not B)
# Here: not (A and not B) = (not A) or B
A, B = True, False
expr1 = not (A and not B)
expr2 = (not A) or B
print('Original == Simplified:', expr1 == expr2)
```

### Question 31: Bitwise Parity Check (Odd/Even)
**Prompt:** Given integer `n`, write an expression using `&` to check whether `n` is odd or even without `%`.


```python
n = 27
# If lowest bit is 1, number is odd
is_odd = (n & 1) == 1
print(f'{n} is odd?', is_odd)
```

### Question 32: Bitwise Left Shift as Power of 2 Multiplier
**Prompt:** Compute the value of `1 << 5` and explain the relationship between left-shift `<< k` and multiplication by $2^k$.


```python
# x << k is mathematically equivalent to x * (2 ** k)
result = 1 << 5
print('1 << 5 =', result)  # 32
```

### Question 33: Bitwise Right Shift as Power of 2 Floor Division
**Prompt:** Compute the value of `64 >> 3` and explain the relationship between right-shift `>> k` and floor division by $2^k$.


```python
# x >> k is mathematically equivalent to x // (2 ** k)
result = 64 >> 3
print('64 >> 3 =', result)  # 64 // 8 = 8
```

### Question 34: In-Place Swap Using XOR
**Prompt:** Swap two integers `a = 15` and `b = 27` in-place using exclusively the `^` (XOR) operator.


```python
a = 15
b = 27
print(f'Before: a={a}, b={b}')

a = a ^ b
b = a ^ b
a = a ^ b

print(f'After:  a={a}, b={b}')
```

### Question 35: XOR Identity Properties
**Prompt:** Predict the output of `x ^ x` and `x ^ 0` for any integer `x`.


```python
x = 42
print('x ^ x =', x ^ x)  # Always 0
print('x ^ 0 =', x ^ 0)  # Always x
```

### Question 36: Single Non-Repeating Number Pattern
**Prompt:** Given an array where every element appears twice except one, explain what expression accumulator pattern solves this in $O(1)$ space.


```python
nums = [4, 1, 2, 1, 2]
unique = 0
for num in nums:
    unique ^= num
print('Single unique element:', unique)
```

### Question 37: Power of Two Bit Trick (`n & (n - 1)`)
**Prompt:** Write an expression using bitwise operators to determine if a positive integer `n` is a power of 2.


```python
# A power of 2 in binary has exactly one bit set (e.g. 16 = 10000, 15 = 01111)
# n & (n - 1) clears the lowest set bit!
n = 16
is_power_of_two = (n > 0) and (n & (n - 1) == 0)
print(f'Is {n} a power of 2?', is_power_of_two)
```

### Question 38: Set the k-th Bit
**Prompt:** Write an expression to set the $k$-th bit of an integer `mask` to `1`.


```python
mask = 0b0000  # 0
k = 3          # 3rd bit (0-indexed)
mask = mask | (1 << k)
print(f'Mask after setting bit {k}: {bin(mask)} ({mask})')
```

### Question 39: Clear the k-th Bit
**Prompt:** Write an expression to clear (set to 0) the $k$-th bit of an integer `mask`.


```python
mask = 0b1111  # 15
k = 2
# Using bitwise NOT (~) and AND (&)
mask = mask & ~(1 << k)
print(f'Mask after clearing bit {k}: {bin(mask)} ({mask})')
```

### Question 40: Toggle the k-th Bit
**Prompt:** Write an expression to toggle (flip) the $k$-th bit of an integer `mask`.


```python
mask = 0b1010  # 10
k = 1
# Using XOR (^)
mask = mask ^ (1 << k)
print(f'Mask after toggling bit {k}: {bin(mask)} ({mask})')
```

### Question 41: Membership Operator (`in`)
**Prompt:** Predict the output: `3 in [1, 2, 3]` vs `'a' in 'cat'`.


```python
print('3 in [1, 2, 3]:', 3 in [1, 2, 3])
print("'a' in 'cat':", 'a' in 'cat')
```

### Question 42: Dictionary Membership Semantics
**Prompt:** Predict the output and explain dictionary lookup: `'name' in {'name': 'Alice', 'age': 25}` vs `'Alice' in {'name': 'Alice', 'age': 25}`.


```python
d = {'name': 'Alice', 'age': 25}
# By default, 'in' checks dict keys, NOT values!
print("'name' in d:", 'name' in d)
print("'Alice' in d:", 'Alice' in d)
print("'Alice' in d.values():", 'Alice' in d.values())
```

### Question 43: In-Place `+=` vs Reassignment `+` on Lists
**Prompt:** What is the difference between `a = a + [1]` and `a += [1]` when `a` is a mutable list referenced by another variable `b = a`?


```python
# Case 1: + creates a new object
a = [1, 2]
b = a
a = a + [3]
print('After a = a + [3]:', 'a =', a, '| b =', b)

# Case 2: += mutates in place via extend()
a = [1, 2]
b = a
a += [3]
print('After a += [3]:    ', 'a =', a, '| b =', b)
```

### Question 44: Exponentiation Associativity (Right-to-Left)
**Prompt:** Evaluate the operator precedence: `2 ** 3 ** 2`. Does exponentiation evaluate left-to-right or right-to-left?


```python
# Exponentiation ** is right-associative!
# Evaluates as: 2 ** (3 ** 2) = 2 ** 9 = 512 (NOT (2 ** 3) ** 2 = 64)
result = 2 ** 3 ** 2
print('2 ** 3 ** 2 =', result)
```

### Question 45: Unary Negation vs Exponentiation
**Prompt:** Predict the output: `-3 ** 2` vs `(-3) ** 2`.


```python
# ** has higher precedence than unary negation (-)
# -3 ** 2 = -(3 ** 2) = -9
print('-3 ** 2 =', -3 ** 2)
print('(-3) ** 2 =', (-3) ** 2)
```

### Question 49: Logical NOT vs Comparison Precedence
**Prompt:** Predict the evaluation of: `not 5 == 5` vs `(not 5) == 5`.


```python
# Comparison (==) has higher precedence than 'not'
# not 5 == 5 -> not (5 == 5) -> not True -> False
print('not 5 == 5:', not 5 == 5)
# (not 5) == 5 -> False == 5 -> False
print('(not 5) == 5:', (not 5) == 5)
```

### Question 50: PEMDAS Arithmetic Evaluation Order
**Prompt:** Evaluate: `100 + 200 / 10 * 2 - 5`. Trace each step according to standard PEMDAS/BODMAS rules.


```python
# / and * have same precedence, evaluated left to right:
# 1. 200 / 10 = 20.0
# 2. 20.0 * 2 = 40.0
# 3. 100 + 40.0 = 140.0
# 4. 140.0 - 5 = 135.0
result = 100 + 200 / 10 * 2 - 5
print('Result:', result)
```
