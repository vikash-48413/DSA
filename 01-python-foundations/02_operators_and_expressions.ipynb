## 1. Implementation From Scratch (Pure Python)
Implement the core concept below without using high-level external libraries.

```python
def is_power_of_two(n: int) -> bool:
    # Bitwise trick: n & (n - 1) == 0 for powers of 2
    return n > 0 and (n & (n - 1)) == 0

print('Is 16 power of 2?', is_power_of_two(16))
print('Is 18 power of 2?', is_power_of_two(18))
```

## 2. Unit Testing & Edge Case Verification
Run the test assertions below to verify correctness of your from-scratch implementation.

```python
assert is_power_of_two(1) == True
assert is_power_of_two(2) == True
assert is_power_of_two(16) == True
assert is_power_of_two(14) == False
print('Bitwise tests passed!')
```
