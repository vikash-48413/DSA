## 1. Implementation From Scratch (Pure Python)
Implement the core concept below without using high-level external libraries.

```python
def handle_agent_action(action: dict) -> str:
    match action:
        case {'type': 'search', 'query': query}:
            return f'Searching web for: {query}'
        case {'type': 'calculator', 'expr': expr}:
            return f'Evaluating expression: {expr}'
        case {'type': 'finish', 'output': out}:
            return f'Final Output: {out}'
        case _:
            return 'Unknown action'

print(handle_agent_action({'type': 'search', 'query': 'FAANG DSA'}))
```
