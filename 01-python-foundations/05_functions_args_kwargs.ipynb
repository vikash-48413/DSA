## 1. Implementation From Scratch (Pure Python)
Implement the core concept below without using high-level external libraries.

```python
def build_model_config(model_name: str, *layers: int, **hyperparams) -> dict:
    return {
        'model_name': model_name,
        'architecture': list(layers),
        'hyperparams': hyperparams
    }

cfg = build_model_config('TransformerBlock', 512, 512, 1024, lr=0.001, dropout=0.1)
print('Model Config:', cfg)
```

## 2. Unit Testing & Edge Case Verification
Run the test assertions below to verify correctness of your from-scratch implementation.

```python
assert cfg['model_name'] == 'TransformerBlock'
assert cfg['architecture'] == [512, 512, 1024]
assert cfg['hyperparams']['lr'] == 0.001
print('Functions test passed!')
```

## 3. Standard Library / Benchmark Comparison
Explore how Python standard library or standard ML packages handle this concept efficiently.

```python
from functools import partial
# Creating partial functions for fixed hyperparameter configs
def train_epoch(model, lr, batch_size):
    return f'Training {model} with lr={lr}, batch_size={batch_size}'

train_gpt = partial(train_epoch, 'GPT-3', lr=1e-4)
print(train_gpt(batch_size=32))
```
