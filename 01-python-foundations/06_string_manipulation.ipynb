## 1. Implementation From Scratch (Pure Python)
Implement the core concept below without using high-level external libraries.

```python
def clean_and_tokenize(text: str) -> list[str]:
    # Normalize to lowercase and remove non-alphanumeric
    cleaned = ''.join(c.lower() if c.isalnum() or c.isspace() else ' ' for c in text)
    return cleaned.split()

print(clean_and_tokenize('Hello, World! AI Engineering 2026.'))
```

## 2. Unit Testing & Edge Case Verification
Run the test assertions below to verify correctness of your from-scratch implementation.

```python
assert clean_and_tokenize('Hello World!') == ['hello', 'world']
print('String manipulation test passed!')
```
