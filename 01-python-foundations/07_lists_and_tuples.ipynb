## 1. Implementation From Scratch (Pure Python)
Implement the core concept below without using high-level external libraries.

```python
def reverse_in_place(arr: list) -> None:
    left, right = 0, len(arr) - 1
    while left < right:
        arr[left], arr[right] = arr[right], arr[left]
        left += 1
        right -= 1

nums = [1, 2, 3, 4, 5]
reverse_in_place(nums)
print('Reversed:', nums)
```

## 3. Standard Library / Benchmark Comparison
Explore how Python standard library or standard ML packages handle this concept efficiently.

```python
import sys
# Compare memory consumption of list vs tuple
tpl = (1, 2, 3, 4, 5)
lst = [1, 2, 3, 4, 5]
print('Tuple size:', sys.getsizeof(tpl), 'bytes')
print('List size:', sys.getsizeof(lst), 'bytes')
```
