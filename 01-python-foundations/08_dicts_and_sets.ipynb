## 1. Implementation From Scratch (Pure Python)
Implement the core concept below without using high-level external libraries.

```python
from collections import defaultdict

def group_anagrams(words: list[str]) -> list[list[str]]:
    anagram_map = defaultdict(list)
    for word in words:
        count = [0] * 26
        for char in word:
            count[ord(char) - ord('a')] += 1
        anagram_map[tuple(count)].append(word)
    return list(anagram_map.values())

print(group_anagrams(['eat', 'tea', 'tan', 'ate', 'nat', 'bat']))
```

## 2. Unit Testing & Edge Case Verification
Run the test assertions below to verify correctness of your from-scratch implementation.

```python
res = group_anagrams(['eat', 'tea'])
assert len(res) == 1 and len(res[0]) == 2
print('Group anagrams test passed!')
```
