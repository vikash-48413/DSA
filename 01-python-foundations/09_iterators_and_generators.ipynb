## 1. Implementation From Scratch (Pure Python)
Implement the core concept below without using high-level external libraries.

```python
def stream_batches(data: list, batch_size: int):
    for i in range(0, len(data), batch_size):
        yield data[i:i + batch_size]

data_stream = list(range(10))
for batch in stream_batches(data_stream, 3):
    print('Batch:', batch)
```

## 2. Unit Testing & Edge Case Verification
Run the test assertions below to verify correctness of your from-scratch implementation.

```python
gen = stream_batches([1, 2, 3, 4], 2)
assert next(gen) == [1, 2]
assert next(gen) == [3, 4]
print('Generator test passed!')
```

## 3. Standard Library / Benchmark Comparison
Explore how Python standard library or standard ML packages handle this concept efficiently.

```python
import itertools
# Using itertools for efficient streaming combinations
print(list(itertools.islice(stream_batches(range(100), 10), 3)))
```
