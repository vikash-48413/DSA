## 1. Implementation From Scratch (Pure Python)
Implement the core concept below without using high-level external libraries.

```python
class TextDataset:
    def __init__(self, texts: list[str]):
        self.texts = texts
        
    def __len__(self) -> int:
        return len(self.texts)
        
    def __getitem__(self, idx: int) -> dict:
        text = self.texts[idx]
        return {'text': text, 'length': len(text)}

ds = TextDataset(['Hello World', 'AI Engineer'])
print('Length:', len(ds))
print('Item 0:', ds[0])
```

## 2. Unit Testing & Edge Case Verification
Run the test assertions below to verify correctness of your from-scratch implementation.

```python
assert len(ds) == 2
assert ds[1]['text'] == 'AI Engineer'
print('OOP dunder method test passed!')
```
