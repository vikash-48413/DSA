## Q001 — Intermediate

**Question:** Find the second-largest digit in a number without sorting.

**Pattern:** `Two-variable tracking`

**Your solution:**


```python
# Write your solution here
n = 8765568905
max1 = -1 # largest
max2 = -1 # second largest

while n > 0:
    d = n % 10
    if d > max1:
        max2 = max1
        max1 = d
    elif d > max2:
        max2 = d
    n = n // 10

print(max2)
```

## Q002 — Intermediate

**Question:** Find the second-smallest digit without sorting.

**Pattern:** `Min tracking`

**Your solution:**


```python
# Write your solution here
n = 92876237
min1 = 9 # smallest
min2 = 9 # second smallest

while n > 0:
    d = n % 10
    if d < min1:
        min2 = min1
        min1 = d
    elif d > min2 and d < min1:
        min2 = d
    n = n //10

print(min2)
```

## Q003 — Intermediate

**Question:** Find the first digit that appears twice in a number.

**Pattern:** `State tracking`

**Your solution:**


```python
# Write your solution here
n = 8756567890
seen = set()
first_duplicate = -1

while n > 0:
    d = n % 10
    n = n //10

    if d in seen:
        first_duplicate = d
        break
    seen.add(d)

print(first_duplicate)
```
