## 1. Implementation From Scratch (Pure Python)
Implement the core concept below without using high-level external libraries.

```python
class HashSet:
    def __init__(self, capacity=16):
        self.capacity = capacity
        self.buckets = [[] for _ in range(self.capacity)]
        
    def _hash(self, key):
        return hash(key) % self.capacity
        
    def add(self, key):
        idx = self._hash(key)
        if key not in self.buckets[idx]:
            self.buckets[idx].append(key)
            
    def contains(self, key) -> bool:
        idx = self._hash(key)
        return key in self.buckets[idx]

hs = HashSet()
hs.add('apple')
hs.add('banana')
print('Contains apple?', hs.contains('apple'))
print('Contains cherry?', hs.contains('cherry'))
```

## 2. Unit Testing & Edge Case Verification
Run the test assertions below to verify correctness of your from-scratch implementation.

```python
assert hs.contains('apple') == True
assert hs.contains('cherry') == False
print('HashSet test passed!')
```
