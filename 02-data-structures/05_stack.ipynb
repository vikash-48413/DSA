## 1. Implementation From Scratch (Pure Python)
Implement the core concept below without using high-level external libraries.

```python
class Stack:
    def __init__(self):
        self.items = []
        
    def push(self, item):
        self.items.append(item)
        
    def pop(self):
        if self.is_empty():
            raise IndexError('Pop from empty stack')
        return self.items.pop()
        
    def peek(self):
        if self.is_empty():
            return None
        return self.items[-1]
        
    def is_empty(self) -> bool:
        return len(self.items) == 0

st = Stack()
st.push(10)
st.push(20)
print('Peek:', st.peek())
print('Pop:', st.pop())
```

## 2. Unit Testing & Edge Case Verification
Run the test assertions below to verify correctness of your from-scratch implementation.

```python
assert st.peek() == 10
assert st.pop() == 10
assert st.is_empty() == True
print('Stack test passed!')
```
