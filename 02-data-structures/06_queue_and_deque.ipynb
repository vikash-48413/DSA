## 1. Implementation From Scratch (Pure Python)
Implement the core concept below without using high-level external libraries.

```python
class Queue:
    def __init__(self):
        self.items = []
        
    def enqueue(self, item):
        self.items.append(item)
        
    def dequeue(self):
        if self.is_empty():
            raise IndexError('Dequeue from empty queue')
        return self.items.pop(0) # O(n) naive list implementation
        
    def is_empty(self) -> bool:
        return len(self.items) == 0

q = Queue()
q.enqueue('token_1')
q.enqueue('token_2')
print('Dequeued:', q.dequeue())
```

## 2. Unit Testing & Edge Case Verification
Run the test assertions below to verify correctness of your from-scratch implementation.

```python
assert q.dequeue() == 'token_2'
assert q.is_empty() == True
print('Queue test passed!')
```

## 3. Standard Library / Benchmark Comparison
Explore how Python standard library or standard ML packages handle this concept efficiently.

```python
from collections import deque
# O(1) append and popleft operations
d = deque(['token_1', 'token_2'])
print('Dequeued O(1):', d.popleft())
```
