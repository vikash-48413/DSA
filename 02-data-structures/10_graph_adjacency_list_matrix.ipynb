## 1. Implementation From Scratch (Pure Python)
Implement the core concept below without using high-level external libraries.

```python
class Graph:
    def __init__(self):
        self.adj_list = {}
        
    def add_edge(self, u, v, weight=1.0, directed=False):
        if u not in self.adj_list:
            self.adj_list[u] = []
        if v not in self.adj_list:
            self.adj_list[v] = []
        self.adj_list[u].append((v, weight))
        if not directed:
            self.adj_list[v].append((u, weight))

g = Graph()
g.add_edge('TokenA', 'TokenB', 0.85)
print('Adjacency List:', g.adj_list)
```

## 2. Unit Testing & Edge Case Verification
Run the test assertions below to verify correctness of your from-scratch implementation.

```python
assert len(g.adj_list['TokenA']) == 1
print('Graph representation test passed!')
```
