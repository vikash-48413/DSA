## 1. Implementation From Scratch (Pure Python)
Implement the core concept below without using high-level external libraries.

```python
def transpose_matrix(matrix: list[list[int]]) -> list[list[int]]:
    rows, cols = len(matrix), len(matrix[0])
    return [[matrix[r][c] for r in range(rows)] for c in range(cols)]

mat = [[1, 2, 3], [4, 5, 6]]
print('Transposed:', transpose_matrix(mat))
```

## 2. Unit Testing & Edge Case Verification
Run the test assertions below to verify correctness of your from-scratch implementation.

```python
assert transpose_matrix(mat) == [[1, 4], [2, 5], [3, 6]]
print('Matrix transpose test passed!')
```
