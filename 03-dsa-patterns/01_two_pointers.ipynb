## 1. Implementation From Scratch (Pure Python)
Implement the core concept below without using high-level external libraries.

```python
def two_sum_sorted(numbers: list[int], target: int) -> list[int]:
    left, right = 0, len(numbers) - 1
    while left < right:
        curr_sum = numbers[left] + numbers[right]
        if curr_sum == target:
            return [left + 1, right + 1]
        elif curr_sum < target:
            left += 1
        else:
            right -= 1
    return []

print('Two sum sorted indices:', two_sum_sorted([2, 7, 11, 15], 9))
```

## 2. Unit Testing & Edge Case Verification
Run the test assertions below to verify correctness of your from-scratch implementation.

```python
assert two_sum_sorted([2, 7, 11, 15], 9) == [1, 2]
print('Two Pointers test passed!')
```
