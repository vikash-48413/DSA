## 1. Implementation From Scratch (Pure Python)
Implement the core concept below without using high-level external libraries.

```python
def max_sum_subarray_k(nums: list[int], k: int) -> int:
    if len(nums) < k:
        return 0
    window_sum = sum(nums[:k])
    max_sum = window_sum
    for i in range(k, len(nums)):
        window_sum += nums[i] - nums[i - k]
        max_sum = max(max_sum, window_sum)
    return max_sum

print('Max sum subarray of size 3:', max_sum_subarray_k([2, 1, 5, 1, 3, 2], 3))
```

## 2. Unit Testing & Edge Case Verification
Run the test assertions below to verify correctness of your from-scratch implementation.

```python
assert max_sum_subarray_k([2, 1, 5, 1, 3, 2], 3) == 9
print('Sliding window test passed!')
```
