## 1. Implementation From Scratch (Pure Python)
Implement the core concept below without using high-level external libraries.

```python
class NumArray:
    def __init__(self, nums: list[int]):
        self.prefix = [0] * (len(nums) + 1)
        for i in range(len(nums)):
            self.prefix[i + 1] = self.prefix[i] + nums[i]
            
    def sum_range(self, left: int, right: int) -> int:
        return self.prefix[right + 1] - self.prefix[left]

na = NumArray([-2, 0, 3, -5, 2, -1])
print('Sum range (0, 2):', na.sum_range(0, 2))
```

## 2. Unit Testing & Edge Case Verification
Run the test assertions below to verify correctness of your from-scratch implementation.

```python
assert na.sum_range(0, 2) == 1
assert na.sum_range(2, 5) == -1
print('Prefix sum test passed!')
```
