# Algorithmic Pattern: Breadth-First Search (BFS)
## FAANG & AI Engineering Practice Notebook

### FAANG Pattern Context
Level-order tree and graph traversal using a queue to find shortest paths in unweighted graphs.

### AI/ML Engineering Mapping

## 1. Implementation From Scratch (Pure Python)
Implement the core concept below without using high-level external libraries.

```python
from collections import deque

def min_steps_bfs(graph: dict, start: str, target: str) -> int:
    queue = deque([(start, 0)])
    visited = {start}
    while queue:
        curr, dist = queue.popleft()
        if curr == target:
            return dist
        for neighbor in graph.get(curr, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, dist + 1))
    return -1

g = {'A': ['B', 'C'], 'B': ['D'], 'C': ['E'], 'D': ['target'], 'E': []}
print('Min steps to target:', min_steps_bfs(g, 'A', 'target'))
```

## 2. Unit Testing & Edge Case Verification
Run the test assertions below to verify correctness of your from-scratch implementation.

```python
assert min_steps_bfs(g, 'A', 'target') == 3
print('BFS test passed!')
```
