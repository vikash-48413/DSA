## 1. Implementation From Scratch (Pure Python)
Implement the core concept below without using high-level external libraries.

```python
def merge_intervals(intervals: list[list[int]]) -> list[list[int]]:
    if not intervals:
        return []
    intervals.sort(key=lambda x: x[0])
    merged = [intervals[0]]
    for curr in intervals[1:]:
        prev = merged[-1]
        if curr[0] <= prev[1]:
            prev[1] = max(prev[1], curr[1])
        else:
            merged.append(curr)
    return merged

print('Merged intervals:', merge_intervals([[1, 3], [2, 6], [8, 10], [15, 18]]))
```
