## 1. Implementation From Scratch (Pure Python)
Implement the core concept below without using high-level external libraries.

```python
class TrieNode:
    def __init__(self):
        self.children = {}
        self.is_end = False

class AutocompleteTrie:
    def __init__(self):
        self.root = TrieNode()
        
    def insert(self, word):
        curr = self.root
        for c in word:
            curr = curr.children.setdefault(c, TrieNode())
        curr.is_end = True
        
    def get_suggestions(self, prefix: str) -> list[str]:
        curr = self.root
        for c in prefix:
            if c not in curr.children:
                return []
            curr = curr.children[c]
            
        res = []
        def dfs(node, path):
            if node.is_end:
                res.append(prefix + ''.join(path))
            for char, next_node in node.children.items():
                dfs(next_node, path + [char])
                
        dfs(curr, [])
        return res

ac = AutocompleteTrie()
for w in ['attention', 'attend', 'attribute', 'transformer']:
    ac.insert(w)
print('Suggestions for "att":', ac.get_suggestions('att'))
```
