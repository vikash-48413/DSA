## 1. Implementation From Scratch (Pure Python)
Implement the core concept below without using high-level external libraries.

```python
class DSU:
    def __init__(self, n):
        self.parent = list(range(n))
        
    def find(self, i):
        if self.parent[i] == i:
            return i
        self.parent[i] = self.find(self.parent[i]) # Path compression
        return self.parent[i]
        
    def union(self, i, j):
        root_i = self.find(i)
        root_j = self.find(j)
        if root_i != root_j:
            self.parent[root_i] = root_j
            return True
        return False

dsu = DSU(5)
dsu.union(0, 1)
dsu.union(1, 2)
print('0 and 2 connected?', dsu.find(0) == dsu.find(2))
```

## 2. Unit Testing & Edge Case Verification
Run the test assertions below to verify correctness of your from-scratch implementation.

```python
assert dsu.find(0) == dsu.find(2)
assert dsu.find(0) != dsu.find(3)
print('DSU test passed!')
```
